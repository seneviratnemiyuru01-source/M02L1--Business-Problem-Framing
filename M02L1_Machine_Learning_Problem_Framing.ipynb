# M02L1 – Machine Learning Problem Framing

## Step 1: Business Request

### Organization
Santa Monica College is a community college that provides educational services to students and employs faculty and staff who depend on campus technology. Its IT support function handles technology-related requests including account access, device issues, and network connectivity.

### Business Request
When a new IT support ticket arrives, support staff must decide whether the issue can be handled by general support or should be escalated to a specialist. The proposed request is to identify tickets likely to need specialist escalation so they can be routed appropriately.

## Step 2: ML Problem Framing

**Type:** Binary classification.

**Input:** One fresh submitted IT support ticket.

**Features:**
1. Ticket category.
2. Urgency or priority.
3. Device type.
4. Operating system or platform.
5. Error code or error category.
6. Short issue description.
7. Similar past tickets and their escalation outcomes.

**Output:** Probability that the ticket will require specialist escalation.

**Decision:** If the predicted probability exceeds the selected threshold, route the ticket to a specialist. Otherwise, stick to general IT support.

### Threshold Prediction

My prediction: 0.5

My reasoning: I predict a threshold near 0.5 because I initially expect the costs of unnecessarily escalating a ticket and losing a ticket that requires specialist support to be balanced.

## Step 3: Compute the Threshold From Money

The decision rule is:

Act when:

p × False Negative Cost > False Positive Cost

Therefore:

Threshold = False Positive Cost / False Negative Cost

For this example, the following are estimates:

- False positive: $15 for unnecessarily escalating a ticket.

- False negative: $60 for failing to escalate a ticket that genuinely needs specialist support.

```python
# Step 3: Calculate the decision threshold

false_positive_cost = 15.0
false_negative_cost = 60.0

threshold = false_positive_cost / false_negative_cost

print("False Positive Cost: $", false_positive_cost)
print("False Negative Cost: $", false_negative_cost)
print("Threshold Calculation:", false_positive_cost, "/", false_negative_cost)
print("Computed Threshold:", threshold)
print("Percentage:", threshold * 100, "%")
```

## Actual Threshold Result

The computed threshold is 0.25 (25%), compared with my original prediction of 0.5.

The calculated threshold is lower because the estimated cost of missing a ticket that requires specialist support (60) is four times greater than the cost of unnecessarily escalating a ticket (15).

## Step 4: Evaluation Metric

The proposed primary metric is recall for the specialist-escalation class.

Recall measures how many of the tickets that genuinely need specialist escalation are correctly identified. This is important because missing a ticket that requires specialist support is estimated to cost more than unnecessarily escalating a ticket.

## Step 5: Data Requirements

The model would require historical IT support tickets with reliable labels depicting whether each ticket needed specialist escalation.

Potential data fields include ticket category, priority, device type, operating system, error information, issue description, and final escalation outcome.

A starting point would be one to two years of historical tickets, depending on the number of examples available.

The likely source would be the organization's IT service desk.

**Weakest point:** Historical escalation labels may be inconsistent because different staff members may make different decisions about similar tickets. A model could learn those inconsistent decisions instead of the actual requirement for specialist expertise.
